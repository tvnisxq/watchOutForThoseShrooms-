```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    classification_report, confusion_matrix,
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, roc_curve
)
print("Imports ran succesfully!")
```

```python
# Setting the visualization ground

# Set teh visual tone for all matplotlib plots to seaborn-darkgrid style
plt.style.use('seaborn-v0_8-darkgrid')
# Set the default figure size for all plots to 12 inches wide by 6 inches tall
plt.rcParams['figure.figsize'] = (12, 6)
# Set the default font size for all text in plots to 12 points 
plt.rcParams['font.size'] = 12

print("Visualization Grounds configured successfully!")
print('\nLets catch some killer shrooms!')
```

```python
# Reading the mushrooms profile: Analyzing Features

# Read the mushroom dataset from csv file and store it in a pandas DataFrame
df = pd.read_csv('data/mushrooms.csv')
print("Data Examined succesfully!")

print('='*50)
print("ANALYSIS BEGINS")
print('='*50)
# Print the total number of mushroom records in the dataset with comma formatting
print(f"Total Number of shrooms: {len(df): ,}")
# Print the number of features (columns) in the dataset
print(f"Features collected per shrooms: {len(df.columns)}")
print("\nFeatures are Listed Below:")

# Loop through each column name with index, starting count from 1
for i, col in enumerate(df.columns, 1):
    # Print the index number and column name in formatted output
    print(f" {i:2d}. {col}")
```

```python
# Let's lineup First 10 shrooms
print('='*50)
print("First 10 shrooms lineup..")
print('='*50)
# Display the first 10 rows of the dataset as a formatted table
print(df.head(10))
```

```python
# 'p' -> poisonous(killer shrooms)
# 'e' -> edible

# Count the number of mushrooms in each class ('e' or 'p') 
class_counts = df['class'].value_counts()
# Calculate the percentage distribution of each class
class_percentages = df['class'].value_counts(normalize=True) * 100

print('='*50)
print("Mushroom Classification Breakdown")

# Loop through each class unique class label in the class_counts
for class_label in class_counts.index:
    # Get the count of mushrooms for the current class 
    count = class_counts[class_label]
    # Get the percentage of mushrooms for the current class
    percentage = class_percentages[class_label]

    # Checks if the class label is 'e' (edible)
    if class_label == 'e':
        # Assign descriptive name "Edible" for display
        class_name = "Edible"
    # If class label is not 'e', it must be 'p'(poisonous)
    else:
        # Assigfn descriptive name "Poisonous" for display
        class_name = "Poisonous"
    
    # Print the class name, count with comma formatting, and percentage 
    print(f"{class_name}: {count:,} mushrooms ({percentage:.2f}%)")

    print(f"\nTotal Shrooms: {len(df):,}")
```

```python
# Visualizing the distributions
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Bar Chart
colors = ['green', 'red'] # Green for safe, red for danger
# Create a bar chart on the first subplot (axes[0])
# X-axis labels: 'Edible' and 'Poisonous
# Y-axis values: counts of edible and poisonous shrooms from class_counts dictionary
bars = axes[0].bar(['Edible', 'Poisonous'],
            [class_counts['e'], class_counts['p']],
            color=colors, edgecolor='black', linewidth=2)
# Set the label for the Y-axis
axes[0].set_ylabel('Number of Mushrooms', fontsize=14)
# Set the lable for the X-axis
axes[0].set_xlabel('Mushroom Classification', fontsize=14)
# Set the title for the subpplot 1
axes[0].set_title("Edible V/s Poisonous", fontweight='bold')
for bar, count in zip(bars, [class_counts['e'], class_counts['p']]):
    axes[0].text(bar.get_x() + bar.get_width()/2, bar.get_height() + 50,
                 f"{count:,},", ha='center', fontsize=12, fontweight='bold')


# Pie Chart
# Create a pie chart on the second subplot (axes[1])
# First list: the values (counts) for each slice 
# Lables: the names to display for each slice ('Edible' and 'Poisonous')
# colors: apply the same color scheme (green and red)
# autopct='%1.1f%%': automatically display percentages on each slice with 1 decimal place
axes[1].pie([class_counts['e'], class_counts['p']],
            labels=['Edible', 'Poisonous'],
            explode=(0, 0.05), shadow=True,
            textprops={'fontsize':14, 'fontweight':'bold',},
            colors=colors,autopct='%1.1f%%')
# Set the title for the pie chart
axes[1].set_title('Probability of surviving random mushroom selection', fontweight='bold')


# Adjust the sublots to prevent overlap
plt.tight_layout()
# Display the figure with both plots
plt.show()
```
