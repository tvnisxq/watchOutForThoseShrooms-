```python
# Reading the mushrooms profile: Analyzing Features

# Read the mushroom dataset from csv file and store it in a pandas DataFrame
df = pd.read_csv('data/mushrooms.csv')
print("Data Examined succesfully!")

print('='*50)
print("ANALYSIS BEGINS")
print('='*50)
# Print the total number of mushroom records in the dataset with comma formatting
print(f"Total Number of shrooms: {len(df): ,}")
# Print the number of features (columns) in the dataset
print(f"Features collected per shrooms: {len(df.columns)}")
print("\nFeatures are Listed Below:")

# Loop through each column name with index, starting count from 1
for i, col in enumerate(df.columns, 1):
    # Print the index number and column name in formatted output
    print(f" {i:2d}. {col}")
```

```python
# 'p' -> poisonous(killer shrooms)
# 'e' -> edible

# Count the number of mushrooms in each class ('e' or 'p') 
class_counts = df['class'].value_counts()
# Calculate the percentage distribution of each class
class_percentages = df['class'].value_counts(normalize=True) * 100

print('='*50)
print("Mushroom Classification Breakdown")

# Loop through each class unique class label in the class_counts
for class_label in class_counts.index:
    # Get the count of mushrooms for the current class 
    count = class_counts[class_label]
    # Get the percentage of mushrooms for the current class
    percentage = class_percentages[class_label]

    # Checks if the class label is 'e' (edible)
    if class_label == 'e':
        # Assign descriptive name "Edible" for display
        class_name = "Edible"
    # If class label is not 'e', it must be 'p'(poisonous)
    else:
        # Assigfn descriptive name "Poisonous" for display
        class_name = "Poisonous"
    
    # Print the class name, count with comma formatting, and percentage 
    print(f"{class_name}: {count:,} mushrooms ({percentage:.2f}%)")
```
